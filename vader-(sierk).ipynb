# VADER Sentiment Analysis

```python
import nltk
#nltk.download('vader_lexicon')        # one-time download
from nltk.sentiment.vader import SentimentIntensityAnalyzer
import pandas as pd

vader_model = SentimentIntensityAnalyzer()

def vader_output_to_label(vader_output):
    """
    map vader output e.g.,
    {'neg': 0.0, 'neu': 0.0, 'pos': 1.0, 'compound': 0.4215}
    to one of the following values:
    a) positive float -> 'positive'
    b) 0.0 -> 'neutral'
    c) negative float -> 'negative'
    """
    compound = vader_output['compound']
    # if compound < 0:
    #     return 'negative'
    # elif compound == 0.0:
    #     return 'neutral'
    # elif compound > 0.0:
    #     return 'positive'
    if compound >= 0.05:
        return 'positive'
    elif compound <= -0.05:
        return 'negative'
    else:
        return 'neutral'

# load the gold test set
st_df = pd.read_csv('Sentiment-topic-test.tsv', sep='\t')

# run VADER on every sentence and collect everything into one table
rows = []
for _, r in st_df.iterrows():
    scores = vader_model.polarity_scores(r['text'])
    pred = vader_output_to_label(scores)
    rows.append({
        'sid': r['sentence id'], 'text': r['text'],
        'gold': r['sentiment'], 'pred': pred,
        'neg': scores['neg'], 'neu': scores['neu'],
        'pos': scores['pos'], 'compound': scores['compound'],
        'correct': pred == r['sentiment'],
    })

results = pd.DataFrame(rows)
results          # display the full table
```
